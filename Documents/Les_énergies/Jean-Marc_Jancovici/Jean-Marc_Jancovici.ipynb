# Jean-Marc JANCOVICI,

## Un fervent défenseur de la lutte contre le réchauffement climatique

### 1. Qui est t'il ?

Jean-Marc Jancovici est un ingénieur, enseignant et conférencier français, spécialisé dans les questions d'énergie et de climat. Diplomé de Polytechnique, il crée en "le Bilan carbone" en 2004, puis en 2007, il cofonde "Carbone 4" avec Alain Grandjean et en 2010, le "Think tank The Shift Project". 

##### a. Le bilan carbone : 

Le bilan carbone est un diagnostic qui quantifie l’ensemble des émissions directes et indirectes d’une organisation, d’un produit ou d’un territoire, exprimées en équivalent CO₂ (CO₂e). Il constitue la base de toute stratégie de décarbonation. 

##### b. Carbone 4 : 

Carbone 4 est un cabinet de conseil français, spécialisé dans la stratégie bas‑carbone, l’adaptation au changement climatique et la biodiversité, en vendant des des bilans carbone aux entreprises. 

##### c. Le "Think tank The Shift Project" : 

Groupe de réflexion réunissant des experts français, dédié à la décarbonation de l'économie et à la transition énergétique. Cette association Loi 1901, a pour but de proposer des soutions pour atténuer le changement climatique tout en assurant la transition vers une économie post-carbone.

Il se fait ensuite, surtout connaître par ses conférences de sensibilisation et de vulgarisation sur les thèmes du réchauffement climatique et de la production d'énergie. En 2018, il devient membre du Haut Conseil pour le climat auprès du Premier ministre et écrit une BD intitulé, "Le Monde sans fin".

### 2. Quelle est son actualité ?

##### a. une assemblée citoyenne pour la présidentielle de 2027 :

Il va réunir 150 citoyens, les faire débattre pendant six mois et publier des propositions sur les grands enjeux de société, avec une résistitution en février 2027. Par cette action, Jean-marc Jancovici souhaite remettre les enjeux économiques, énergétique et climatiques au coeur du débat politique.

#####  b. Un nouveau livre sur l'économie Française : 

"The Shift Project" prévoit de publié un livre intitulé : "Réussir à l'ère des crises : un plan robuste pour l'économie Française". Cette ouvrage à pour but de proposer des mesures pour rendre la France moins dépendante des énergie fossiles et plus résiliente face aux crises. 

##### c. Des interventions télévisés sur le pétrole et la transition énergétique : 

- Il était l'invité du jour à Télématin (France 2), le 29 septembre dernier, pour parlet du climat et de son projet d'assemblée citoyene. 

- Le 27 août, il a participé à l'émission "C dans l'air" afin de présenter également son projet d'assemblée citoyenne. 

### 3. Quelles sont ses thèses ?

- Il milite notamment pour la sobriété énergétique

- En faisant référence à la construction de l'A69, il déclare "il ne faut pas construire les nouvelles autoroutes et s’occuper en priorité d’électrifier le transport de marchandises"

- Il faut privilégier le transport férovière et renoncer à l'avion

- Selon lui, la voiture éléctrique n'est pas "propre" car elle ne consomme pas d'éléctricité, elle même issu d'un modèle durable. 

- Le modèle actuel des sociétés occidentales est voué à la décroissance car il repose sur des combustibles fossiles

- Il propose de miser sur l'énergie nucléaire en plus du développement des énergies renouvelables

- Le réchauffement climatique se double, à cause de l'épuisement avancé des énergie fossiles


- En 2030, un pic pétrolier entraineraitla chute de la consommation, l'appovrissement des classes moyennes, la recession de l'économie Mondiale et une forte pression migratoire aux portes de l'Europe. 

### 4. Fait-il l'unanimité ?

Pas vraiment, il est vivement critiquer sur ces positions dites "pronucléaires" et sur ses idéee de l'énergie renouvelables. 

##### 1. 
Certains écologistes comme Marine Tondelier, lui reprochent de sous-estimer le rôle que peuvent jouer l'éolien et le solaire dans la transition énergétique. 

###### 2. 
Il est également critiqué par le Monde, qui déclare "Jean-Marc Jancovici est devenu ces dernières années un des principaux porte-voix de la décroissance en France". 

###### 3.
Certains chercheurs estiments que ses simplifications déstinées aux grand public peuvent conduire à une présentation biaisée de sujets complexes. 


Sinon, beaucoup de partisans du nucléaire le soutiennent, et il est aussi appuyé par "The Shift Project"
