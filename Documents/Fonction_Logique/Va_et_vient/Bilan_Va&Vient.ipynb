# Va et vient réalisé avec Compar IA : 

### Prompt :

Génère le code d'une page HTML avec du CSS et du Javascript pour simuler le fonctionnement d'un éclairage par va et vient électrique. Une ampoule L s'allume si on clique sur un switch A pour l'activer et qu'on ne clique pas sur un switch B ou si on clique sur le switch B et qu'on ne clique pas sur le switch A , sinon l'ampoule L reste éteinte. Fais aussi un logigramme, une table de vérité et un chronogramme. Le tout doit être instructif et interractif.

### Réflexion de l'IA (Claude Sonnet 5.5) :

"use strict";

let A=0,B=0,timer=null,phase=0;
const $=id=>document.getElementById(id);

function xor(a,b){
  return a!==b?1:0;
}

function update(){
  const L=xor(A,B);

  $("btnA").classList.toggle("active",!!A);
  $("btnB").classList.toggle("active",!!B);
  $("btnA").setAttribute("aria-pressed",String(!!A));
  $("btnB").setAttribute("aria-pressed",String(!!B));

  $("posA").textContent="Position "+A;
  $("posB").textContent="Position "+B;

  $("bulb").classList.toggle("on",!!L);
  $("status").classList.toggle("on",!!L);
  $("status").textContent=L?"● Lampe allumée":"○ Lampe éteinte";

  $("explanation").innerHTML=L
    ?"<b style='color:var(--green)'>Sortie L = 1</b><br>A et B sont différents : la lampe est allumée."
    :"<b>Sortie L = 0</b><br>A et B sont identiques : la lampe est éteinte.";

  ["c1","c2","c3","c4","c5","c6"].forEach((id,i)=>{
    $(id).classList.toggle("active",([A===0,A===0,A===1,A===0,A===1,L===1][i]));
  });

  $("leverA").setAttribute("x2","229");
  $("leverA").setAttribute("y2",A?"135":"45");
  $("leverB").setAttribute("x2","334");
  $("leverB").setAttribute("y2",B?"135":"45");

  $("leverA").setAttribute("stroke",A?"#ffc857":"#a0b1ce");
  $("leverB").setAttribute("stroke",B?"#ffc857":"#a0b1ce");

  $("svgBulb").setAttribute("fill",L?"#ffe69a":"#26334c");
  $("svgBulb").setAttribute("stroke",L?"#ffc857":"#8192ae");
  $("bulbCross").setAttribute("stroke",L?"#9a6b00":"#8192ae");

  $("svgState").textContent=`A=${A} · B=${B} · L=${L}`;
  $("logicValues").textContent=`A=${A} · B=${B} · L=${L}`;

  document.querySelectorAll("[data-gate]").forEach(g=>{
    g.classList.toggle("on",
      g.dataset.gate==="L"?!!L:
      g.dataset.gate==="x1"?(A===1&&B===0):
      (A===0&&B===1)
    );
  });

  document.querySelectorAll("[data-wire]").forEach(w=>{
    let n=w.dataset.wire;
    w.classList.toggle("on",
      n==="A"?!!A:
      n==="B"?!!B:
      n==="nA"?!A:
      n==="nB"?!B:
      n==="x1"?(A===1&&B===0):
      n==="x2"?(A===0&&B===1):
      !!L
    );
  });

  document.querySelectorAll("#truth tr").forEach(r=>{
    r.classList.toggle("current",+r.dataset.a===A&&+r.dataset.b===B);
  });
}

$("btnA").addEventListener("click",()=>{
  A=1-A;
  update();
});

$("btnB").addEventListener("click",()=>{
  B=1-B;
  update();
});

$("reset").addEventListener("click",()=>{
  A=0;
  B=0;
  phase=0;
  update();
});

$("step").addEventListener("click",()=>{
  const states=[[0,0],[0,1],[1,1],[1,0]];
  phase=(phase+1)%states.length;
  [A,B]=states[phase];
  update();
});

document.querySelectorAll("#truth tr").forEach(r=>{
  r.addEventListener("click",()=>{
    A=+r.dataset.a;
    B=+r.dataset.b;
    update();
  });
});

  function buildTruthTable() {

    for (var i = 0; i < 4; i++) {

      tr.innerHTML =
        "<td>" + (i + 1) + "</td><td>" + a + "</td><td>" + b + "</td>" +
        "<td>" + l + "</td>" +

  /* ---------- Chronogramme ---------- */
  var CH = { x0: 60, rowH: 70, colW: 60, hiOff: 26, labelX: 15, exp: 8 };
  function drawChrono() {
    var svg = document.getElementById("chrono");
    var rowsL = [0, 1, 2]; // A, B, L
    var cols = Math.max(history.length, 6); // au moins 6 colonnes pour garder une largeur stable
    var width = CH.x0 + cols * CH.colW + 20;
    var height = 40 + rowsL.length * CH.rowH + 50;
    svg.setAttribute("width", width);
    svg.setAttribute("height", height);
    svg.setAttribute("viewBox", "0 0 " + width + " " + height);

    function add(el, attrs, txt) {
      var e = document.createElementNS("http://www.w3.org/2000/svg", el);
      for (var k in attrs) e.setAttribute(k, attrs[k]);
      if (txt != null) e.textContent = txt;
      svg.appendChild(e);
      return e;
    }

    var labels = ["A", "B", "L = A ⊕ B"];
    var names = ["A", "B"];
    var colors = ["#38bdf8", "#a78bfa", "#fbbf24"];

    function pathFor(row, values) {
      // construit un tracé en escalier : rows : y pour la valeur 1 et pour 0
      var high = 40 + row * CH.rowH + CH.hiOff;
      var low = high + 34;
      var d = "M" + CH.x0 + " " + (values[0] ? high : low);
      var x = CH.x0;
      for (var i = 1; i < values.length; i++) {
        x += CH.colW;
        if (values[i] === values[i - 1]) {
          d += " H" + x;
        } else if (values[i] === 1) {
          d += " V" + high + " H" + x;
        } else {
          d += " V" + low + " H" + x;
        }
      }
      return { d: d, high: high, low: low };
    }

    var vals = [history.map(function (h) { return h.A; }),
                history.map(function (h) { return h.B; }),
                history.map(function (h) { return h.A ^ h.B; })];

    rowsL.forEach(function (r) {
      var col = colors[r];
      var p = pathFor(r, vals[r]);
      // graduations de temps
      for (var i2 = 0; i2 <= cols; i2++) {
        var x2 = CH.x0 + i2 * CH.colW;
        add("line", { x1: x2, y1: p.high - 10, x2: x2, y2: p.low + 10, stroke: "#1e293b", "stroke-width": 1 });
        if (r === 2 && i2 < history.length && i2 % 2 === 0) {
          add("text", { x: x2, y: height - 14, fill: "#64748b", "font-size": 11, "text-anchor": "middle",
            "font-family": "Consolas, monospace" }, "t" + i2);
        }
      }
      add("text", { x: CH.labelX, y: p.high + 12, fill: col, "font-size": 15, "font-weight": "bold",
        "font-family": "Consolas, monospace" }, labels[r]);
      add("path", { d: p.d, fill: "none", stroke: col, "stroke-width": 3,
        "stroke-linejoin": "miter", stroke: col,
        filter: r === 2 ? "drop-shadow(0 0 4px " + col + ")" : undefined });

      // niveau "1" / "0" repères
      add("text", { x: CH.labelX + 6, y: p.high + 6, fill: "#475569", "font-size": 10 }, "1");
      add("text", { x: CH.labelX + 6, y: p.low + 10, fill: "#475569", "font-size": 10 }, "0");
    });

    // marqueurs : transitions où L change
    for (var i = 1; i < history.length; i++) {
      var prev = history[i - 1], cur = history[i];
      if ((prev.A ^ prev.B) !== (cur.A ^ cur.B)) {
        var x = CH.x0 + i * CH.colW;
        add("line", { x1: x, y1: 30, x2: x, y2: height - 34, stroke: "#facc15", "stroke-width": 1,
          "stroke-dasharray": "4 4", opacity: 0.6 });
      }
    }
  }

  /* ---------- Init ---------- */
  buildTruthTable();
  render();
})();
