Quelles étaient les raisons de cet hommage rendu par Google ?

- il a inventé le language binaire et google s'appuie sur ce systeme. 

- google utilise aussi les operateurs logique inventé par celui ci : AND OR NOT.

Qu'est-ce qu'une variable booléenne ?

- Il s'agit d'une variable qui ne prend que 2 valeurs comme TRUE FALSE ou 1 et 2

Qu'elles sont les principales fonctions (opérations) booléennes ?

- Fonction AND : TRUE AND TRUE = TRUE / TRUE AND FALSE = FALSE

- Fonction OR : TRUE OR FALSE = TRUE / FALSE OR FALSE = FALSE 

- Fonction NOT : NOT TRUE = FALSE / NOT FALSE = TRUE
